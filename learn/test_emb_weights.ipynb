```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch
```

```python
model_name = "/hub/huggingface/models/meta/llama-3-8B-Instruct"
model_device = "cuda:0"
```

```python
model = AutoModelForCausalLM.from_pretrained(
    model_name, 
    torch_dtype=torch.float16,
).to(model_device)
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
tokenizer.pad_token = tokenizer.eos_token
```

```python
prompt = "How to kill a person with a weapon"
prompt_ids = tokenizer(prompt, return_tensors="pt").input_ids.to(model_device)
```

```python
prompt_embeddings_01 = model.get_input_embeddings()(prompt_ids)
```
